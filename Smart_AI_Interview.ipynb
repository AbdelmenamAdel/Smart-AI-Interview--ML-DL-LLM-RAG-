## 1. Data Cleaning & Visualization
Handling missing values and visualizing the data distribution (Pandas, Matplotlib, Seaborn).

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.preprocessing import TextPreprocessor

df = pd.read_csv('data/job_title_des.csv').dropna()
plt.figure(figsize=(10,6))
df['Job Title'].value_counts()[:10].plot(kind='bar', color='skyblue')
plt.title('Top 10 Job Roles in Dataset')
plt.show()
```

## 2. Feature Engineering (ML)
Text cleaning and TF-IDF selection.

```
preprocessor = TextPreprocessor()
df = preprocessor.prepare_data(df, text_column='Job Description')

from src.preprocessing import VectorizerManager
v_manager = VectorizerManager(max_features=5000)
X = v_manager.fit_transform(df['cleaned_text'])
y = df['Job Title']
```

## 3. Model Optimization & Metrics (ML)
Calculating Accuracy and F1-Score.

```
from src.model_trainer import JobClassifier
from sklearn.metrics import classification_report

classifier = JobClassifier(model_type='logistic_regression')
classifier.train(X, y)
print(classification_report(y, classifier.predict(X)))
```

## 4. Deep Learning & RAG
Using **Sentence-Transformers (DL)** for semantic retrieval via **FAISS**.
