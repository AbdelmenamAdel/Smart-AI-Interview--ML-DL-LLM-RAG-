# Smart AI Interview Project
This notebook outlines the modular workflow for job classification and interview question retrieval.

## 1. Import Modules
We use the modular scripts developed in the `src/` directory.

```
import pandas as pd
from src.preprocessing import TextPreprocessor, VectorizerManager
from src.model_trainer import JobClassifier
from src.interview_logic import InterviewLogic
```

## 2. Load Datasets

```
df = pd.read_csv('data/job_title_des.csv')
df.head()
```

## 3. Preprocessing and Vectorization

```
preprocessor = TextPreprocessor()
df = preprocessor.prepare_data(df, text_column='Job Description')

v_manager = VectorizerManager(max_features=5000)
X = v_manager.fit_transform(df['cleaned_text'])
y = df['Job Title']
```

## 4. Model Training and Evaluation

```
classifier = JobClassifier(model_type='logistic_regression')
classifier.train(X, y)
accuracy, report = classifier.evaluate(X, y)
print(f'Accuracy: {accuracy}')
```

## 5. Inference and Question Retrieval

```
interview_logic = InterviewLogic('data/mcq_questions.csv', 'data/essay_questions.csv')

test_description = "Looking for a Full Stack Developer with React and Node.js skills."
cleaned_test = preprocessor.clean_text(test_description)
X_test = v_manager.transform([cleaned_test])
predicted_role = classifier.predict(X_test)[0]

print(f'Predicted Role: {predicted_role}')
questions = interview_logic.get_questions(predicted_role)
print(interview_logic.format_questions(questions))
```
